```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('adult.csv')
df = df.drop(columns=['education.num','fnlwgt','education','relationship','capital.gain','capital.loss'], errors='ignore')
df = df[['age','workclass','marital.status','occupation','sex','hours.per.week','native.country','race','capital.gain','capital.loss','income']]
df.head()
```

```python
coluna_alvo = 'income'
```

```python
amostra_5k, resto_do_dataset = train_test_split(
    df, 
    train_size=300, 
    stratify=df[coluna_alvo], 
    random_state=42  # A semente garante que você gere a mesma amostra se rodar o script novamente
)
```
